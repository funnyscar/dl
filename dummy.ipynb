```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

r_a = np.linspace(1000, 2500, 150)
r_a_test = 99 # index of the rating to test
r_b = 1800

k = 500

e_a = 1 / (1 + 10 ** ((r_b - r_a) / 400))
print("expected to win?:", e_a)

s_a = 1
r_a_new = r_a + k * (s_a - e_a)
plt.plot(r_a, r_a_new, 'o', c='green')
print("win: ", r_a_new[r_a_test])

s_a = 0.5
r_a_new = r_a + k * (s_a - e_a)
plt.plot(r_a, r_a_new, 'o', c='orange')
print("tie: ", r_a_new[r_a_test])

s_a = 0
r_a_new = r_a + k * (s_a - e_a)
plt.plot(r_a, r_a_new, 'o', c='red')
print("lose: ", r_a_new[r_a_test])


plt.plot(r_a, r_a)
```

```python
import os
import nest_asyncio
nest_asyncio.apply()

from dotenv import load_dotenv
load_dotenv("../autoresearch/.env")

from openevolve import run_evolution, evolve_function
from openevolve.config import Config, LLMModelConfig

config = Config()
config.llm.models = [
    LLMModelConfig(
        name="claude-haiku-4-5-20251001",
        api_key=os.environ["ANTHROPIC_API_KEY"],
        api_base="https://api.anthropic.com/v1/",
    )
]

# Evolve Python functions directly
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j] 
    return arr

result = evolve_function(
    bubble_sort,
    test_cases=[([3,1,2], [1,2,3]), ([5,2,8], [2,5,8])],
    iterations=50,
    config=config,
)
print(f"Evolved sorting algorithm: {result.best_code}")
```
