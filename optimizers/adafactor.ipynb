# adafactor
https://arxiv.org/abs/1804.04235

adam uses too much memory when models expand.


**problem:**

Too many N*M *2 (beta1, beta2) moving avg to keep track of. 

solution:
- use I-divergence: d(p, q)
- minimize and derive analytic solution in lemma 1.
- O(N + M) mem size.
- get algorithm 2.

**problem:**
- algo 2.
- all parameters in model (MLP) are incremented according to ADAM. yields instability problems in figure 1.
- approx dynamically changing step size: $u_{xt}$ ~ $g_{xt} / \sqrt{v_{xt}}$
- rms(u) ~ 1 is good since $v_{xt}$ is supposed to approximate $g^2$.

solution:
- clip when rms gets too high.
- $\hat{U_t} = \frac{U_t}{\max{(1, rms(U_t) / d)}}$

solution:
- decaying beta2. (Reddi et al. 2018)
